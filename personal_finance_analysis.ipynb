## 1. IMPORT LIBRARIES

```python
import pandas as pd

print("Pandas imported successfully!")
```

## 2. READ CSV FILE

```python
df = pd.read_csv("dataset/personal_transactions.csv")

print("First 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())
```

## 3. DATA CLEANING AND BASIC ANALYSIS

```python
# DATA CLEANING AND BASIC ANALYSIS

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Remove duplicate records
df = df.drop_duplicates()

# Remove rows with missing important values
df = df.dropna(
    subset=["Date", "Amount", "Transaction Type"]
)

# Convert Date and Amount
df["Date"] = pd.to_datetime(
    df["Date"], errors="coerce"
)

df["Amount"] = pd.to_numeric(
    df["Amount"], errors="coerce"
)

# Remove invalid records
df = df.dropna(subset=["Date", "Amount"])

# Standardize transaction type
df["Transaction Type"] = (
    df["Transaction Type"].astype(str).str.strip().str.lower()
)

print("\nCleaned Dataset:")
print(df.head())

print("\nDataset Shape:", df.shape)

print("\nTransaction Types:")
print(df["Transaction Type"].value_counts())

print("\nCategories:")
print(df["Category"].unique())
```

## 4.FINANCIAL SUMMARY

```python
# FINANCIAL SUMMARY

total_credit = df.loc[
    df["Transaction Type"] == "credit", "Amount"
].sum()

total_debit = df.loc[
    df["Transaction Type"] == "debit", "Amount"
].sum()

net_difference = total_credit - total_debit

print("========== FINANCIAL SUMMARY ==========")

print("Total Credit:", round(total_credit, 2))
print("Total Debit:", round(total_debit, 2))
print("Net Difference:", round(net_difference, 2))

print("\nTotal Transactions:", len(df))
print("Average Transaction:", round(df["Amount"].mean(), 2))
```

## 5. FIRST DATE AND LAST DATE

```python
print("First Date:", df["Date"].min())
print("Last Date:", df["Date"].max())
```

## 6. CATEGORY-WISE ANALYSIS

```python
# CATEGORY-WISE ANALYSIS

category_analysis = (
    df[df["Transaction Type"] == "debit"]
    .groupby("Category")["Amount"]
    .sum()
    .sort_values(ascending=False)
)

print("Category-wise Debit Analysis:")
print(category_analysis)
```

## 7. EXPORT CLEANED CSV

```python
df.to_csv("cleaned_transactions.csv", index=False)

print("Cleaned CSV file saved successfully!")
```

## 9. BUDGET COMPARISON

```python
budget_df = pd.read_csv("dataset/Budget.csv")

actual_expenses = (
    df[df["Transaction Type"] == "debit"]
    .groupby("Category")["Amount"]
    .sum()
    .reset_index()
)

budget_comparison = pd.merge(
    budget_df,
    actual_expenses,
    on="Category",
    how="left"
)

budget_comparison["Amount"] = budget_comparison["Amount"].fillna(0)

budget_comparison["Difference"] = (
    budget_comparison["Budget"] -
    budget_comparison["Amount"]
)

print("BUDGET VS ACTUAL SPENDING")
print(budget_comparison)
```

## 10. FINAL BUDGET STATUS

```python
budget_comparison["Status"] = budget_comparison["Difference"].apply(
    lambda x: "Within Budget" if x >= 0 else "Over Budget"
)

print("\nFINAL BUDGET STATUS")
print(budget_comparison)
```

## 11. SAVE BUDGET COMPARISON CSV

```python
budget_comparison.to_csv("budget_comparison.csv", index=False)

print("Budget comparison file saved successfully!")
```
